```python
import pandas as pd

veri = {
    'Ogrenci': ['Ahmet', 'Mehmet', 'Ayşe', 'Fatma'],
    'Vize': [70, 50, 85, 90],
    'Final': [80, 60, 95, 85]
}
df = pd.DataFrame(veri)


df['Ortalama'] = (df['Vize'] * 0.4) + (df['Final'] * 0.6)


basarililar = df[df['Ortalama'] > 75]

print("Tüm Öğrenciler ve Notlari:\n", df)
print("\n--- 75 Üstü Alanlar ---:\n", basarililar)

```

```python
import pandas as pd

kirli_veri = {
    'oyun': ['  Fifa ', 'GTA', 'Gow', '  Fifa ', 'Minecraft'],
    'kategori': [' SPOR', 'acik dunya ', 'Hikaye', ' SPOR', None],
    'fiyat': ['1200', '900', None, '1200', '450'],
}

df = pd.DataFrame(kirli_veri)
df['kategori']=df['kategori'].fillna('acik dunya')
df['fiyat']=df['fiyat'].fillna('400')
df['fiyat']=df['fiyat'].astype(int)
df['kategori']=df['kategori'].str.strip().str.lower()
df['oyun']=df['oyun'].str.strip().str.lower()
df=df.drop_duplicates()
print(df.info())
```
